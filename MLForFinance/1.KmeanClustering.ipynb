# K-Means Clustering

Unsupervised learning is a branch of machine learning designed to discover hidden patterns and structures in data without labeled ground truths3. In asset management, these techniques are highly valuable because financial markets are opaque, relationships undergo rapid regime shifts, and labeled data are scarce, expensive, or prone to becoming obsolete4.The most well-known framework for unsupervised learning is clustering.

**K-Means Clustering** is one of the most fundamental and widely used unsupervised learning algorithms. It is designed to discover hidden patterns and structures in numeric datasets without replying on pre-existing labels or ground truths.

The algorithms operates on a compactness cireterion, meaning its goal it to partition data into *k* distinct , non-overlapping groups where points within the same group are as close to each other as possible.


### Core-Mathematical Mechanics

The algorithm uses the concept of **centriods** , which are average(mean) position of all the data points belonging to a particular cluster. This classic K-Means workflows (typcially known as **Lloyd Algorithm**) execute the following interactive steps:

1. **Centriod Intialization**: Select *k* centriods randomly from the datasets:
    $\mu_1, \mu_2, \dots, \mu_k $   

2. **Assignment Step**: For Each individual data point say $x_i$ in the dataset $X$, calculate its euclidian distance of all *k* centriods:
    $$ d(x_i, \mu_j) =||x_i - \mu_j||_2$$

    Assign $x_i$ to the closest centroid's cluster, denoted by $c_i$: 
    $$c_i = \arg\min_j d(x_i, \mu_j)$$


3. **Update Step**: Recompute the position of each centriod $\mu_j$ by calculating the mathematical mean of all data points currently assigned to that cluster:
   
$$\mu_j = \frac{1}{|C_j|} \sum_{x_i \in C_j} x_i$$

4. **Convergence Check**:Repeat the assignment and update steps sequentially. The algorithm stops(converges) when the centriods no longer shift position significantly  between iterations, or when it reached the pre-specified limit of maximum iterations.


### Analytical Trade-Offs5:
- Strengths: It is mathematically simple, computationally efficient, easy to interpret, and highly scalable for large datasets.
- Assumptions: Assumes clusters are numeric, roughly spherical, and similar in size.
- Limitations: Highly sensitive to centroid initialization and outliers. It requires pre-specification of the cluster count $k$, and can only detect clusters that are linearly separable5.

## Simulating a Multi-Regime Financial Dataset

To Replicate realistic asset management conditions, we generate a synthetic dataset representing **three distinct economic regimes** across two variables: **Asset Returns** and the **Volatility Risk Premium(VRP)**:

1. **Regime 1(Normal Expansion)**: Moderate Positive returns and low volatility(representing a standard bull market).
2. **Regime 2(Criss/Sell Off)**: Negative returns and extremly hight volatility(representing a severe bear market or liquidity panic).
3. **Regime 3(High Growth Recovery)**: Strong positive returns and moderate volatility (representing a post-crisis expansion/rebound):





```python
n_samples_per_regime = 100

# Regime 1: Normal Growth (Centered around Return = 1.5%, VRP = -0.5%)
regime1_returns = np.random.normal(loc=1.5, scale=0.8, size=(n_samples_per_regime, 1))
regime1_vrp = np.random.normal(loc=-0.5, scale=0.4, size=(n_samples_per_regime, 1))
regime1 = np.hstack((regime1_returns, regime1_vrp))

# Regime 2: Bear Market/Crisis (Centered around Return = -3.0%, VRP = 2.5%)
regime2_returns = np.random.normal(loc=-3.0, scale=1.5, size=(n_samples_per_regime, 1))
regime2_vrp = np.random.normal(loc=2.5, scale=0.8, size=(n_samples_per_regime, 1))
regime2 = np.hstack((regime2_returns, regime2_vrp))

# Regime 3: High-Growth Recovery (Centered around Return = 4.0%, VRP = -1.2%)
regime3_returns = np.random.normal(loc=4.0, scale=1.0, size=(n_samples_per_regime, 1))
regime3_vrp = np.random.normal(loc=-1.2, scale=0.5, size=(n_samples_per_regime, 1))
regime3 = np.hstack((regime3_returns, regime3_vrp))

# Combine and label
X = np.vstack((regime1, regime2, regime3))
true_labels = np.array(*100 + [8]*100 + [9]*100)

# Wrap in DataFrame
df = pd.DataFrame(X, columns=['Equity_Returns_Factor', 'Volatility_Risk_Premium'])
print(f"Dataset generated with shape: {df.shape}")
df.sample(5)
```
