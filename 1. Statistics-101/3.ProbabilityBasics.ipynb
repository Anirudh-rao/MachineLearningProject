## 3. Random Variables
A random variable is a quantity whose value is uncertain before it is observed. Unlike ordinary variables in mathematics that take fixed values, random variables represent outcomes of random processes that can vary from one observation to another. Random variables connect abstract probability theory and real-world measurements, allowing us to apply mathematical reasoning to uncertain phenomena.

We distinguish between two types of random variables based on the values they can assume:

1. **Discrete random variables** take on countable values, often integers, such as the number of customers arriving at a store, the count of defective items in a batch, or the outcome of rolling a die.
2. **Continuous random variables** can take any value within an interval, such as temperature, time, height, or stock prices. The distinction matters because discrete and continuous random variables require different mathematical tools: probability mass functions for discrete cases and probability density functions for continuous cases.

Consider a simple example: let 
$X$ represent the outcome of rolling a fair six-sided die. This discrete random variable can take values in the set 
${1,2,3,4,5,6}$, each with equal probability 
$1/6$. In contrast, if 
$Y$ represents the time until the next customer arrives at a store, this continuous random variable could take any positive real value, and we describe its behavior using a continuous probability distribution like the exponential distribution.



## 4. Probablity Distribution

A probability distribution describes how probabilities are spread across the possible values of a random variable. It tells us which outcomes are likely and which are rare. Different types of random phenomena follow different patterns, and recognizing these patterns helps us model real-world processes effectively.

For discrete random variables, we use a **probability mass function**(PMF) that assigns probabilities to each possible value. The PMF must satisfy two conditions: 
- all probabilities must be non-negative, and 
- they must sum to one across all possible outcomes. 


For continuous random variables, we use a **probability density function** (PDF) that describes the relative likelihood of values in different regions. Unlike discrete probabilities, the PDF itself does not give probabilities directly. Instead, we integrate the PDF over an interval to find the probability that the variable falls within that range.

Several probability distributions appear repeatedly due to their theoretical properties and empirical relevance. 
- The **normal distribution** (Gaussian distribution) describes many natural phenomena and plays a central role in statistical theory due to the Central Limit Theorem. 
- **The binomial distribution** models the number of successes in a fixed number of independent trials, such as counting how many emails out of 100 are spam.

Choosing the distribution that best describes your data helps you build an appropriate statistical model. Misspecifying the distribution can lead to incorrect inferences, poor predictions, and unreliable uncertainty estimates.

## 5. Calculating Probabilities from PMF's

Once we have defined a probability distribution, we can calculate the probability of specific outcomes or ranges of outcomes. The approach differs depending on whether we're working with discrete or continuous distributions.

1. For **For discrete distributions**, we use the **Probability Mass Function** (PMF), which gives us 
$P(X=x)$, the probability that the random variable equals exactly a specific value. The PMF must satisfy two key properties:

1. $P(X = x) >= 0 for all value of x$(probablities are non-negative)
2. $\sum_{all_x}P(X=x)=1 $( probabilities  sum  to  1)


To find the probability of a range of outcomes, we simply sum the individual probabilities. For example, 
$P(2≤X≤5)=P(X=2)+P(X=3)+P(X=4)+P(X=5)$.

An important consequence is that for continuous distributions,

3. The **The Cumulative Distribution Function (CDF)** provides another useful tool for calculating probabilities. The CDF, denoted 
$F(x)$, gives the probability that the random variable is less than or equal to `x`:

$$ F(x) = P(X <= x)$$

For continuous distributions, the CDF is obtained by integrating the PDF from 
`−∞` to 
`x`:

$$F(x) = \int_{−∞}^{x} f(t) dt $$

In other words, the CDF at any point equals the total area under the PDF curve to the left of that point.

## 6. Expected Value and Variance

While a full probability distribution gives complete information about a random variable, we often need summary statistics that capture the key features with just a few numbers. The two most important summary measures are the **expected value** and **variance**, which describe the center and spread of a distribution.

### a. Expected Value
The expected value (or mean) of a random variable represents its long-run average value if we repeated the random process infinitely many times. For a discrete random variable 
`X` with values $x_1,x_2,\dot,x_n$
​and corresponding probabilities $p_1,p_2,\dot,p_n$, the expected value is:

$$E[X] = \sum_{i=1}^{n}x_{i} * p_{i}$$


**Example: Expected Number of Heads in a Coin flips**:

For a binomial random variable (like counting heads in coin flips), there's a convenient formula: 
$E[X]=n⋅p$, where 
`n` is the number of trials and 
`p` is the probability of success on each trial. If we flip a fair coin 100 times, the expected number of heads is:

$$ E[X] = 100 * 0.5 = 50$$

This means that over many repetitions of 1000 coin flips, the average number of heads will converge to 500.

**Example: Expected Value of a Fair Die Roll**

Consider rolling a fair six-sided die. The random variable 
`X` represents the outcome, which can be any value from `1` to `6`, each with probability `1/6`:


Notice that the expected value of `3.5` is not even a possible outcome of a single die roll! This illustrates an important point: *the expected value represents the long-run average, not necessarily an achievable result*. If you rolled a die 1000 times and averaged all the results, you'd get very close to `3.5`.

For a continuous random variable with probability density function 
$f(x)$, the expected value is:

$$E[X] = \int_{-−∞}^{∞} x * f(x) dx$$

The expected value provides a measure of central tendency, answering the question: *"What value should I expect on average?"* However, knowing only the average doesn't tell us how much variation to expect around that average.

### b.Variance

The variance measures how spread out the distribution is around its expected value. It quantifies the average squared deviation from the mean, giving us a sense of how much uncertainty or variability exists in the random variable. The variance of a random variable 
X is defined as:

$$ Var(X) = E[(X- E[X]^2)] = E[X^2] - (E[X])^2 $$

Where
- $Var(X)$: Variance of the randome variable `X`.
- $E[X]$ : Expected Value(Mean)
- $E[X^2]$: Expected Value of Squared random variable.

A closely related measure is the **standard deviation**, which is simply the square root of the variance: $σ = \sqrt{Var(X)}$. The standard deviation has the advantage of being in the same units as the original random variable, making it more interpretable. For example, if we're measuring heights in centimeters, the standard deviation is also in centimeters, while the variance would be in squared centimeters.

These summary statistics prove invaluable in practice. For example, when comparing investment options, the expected value tells us the average return, while the variance indicates the risk.When designing systems, knowing the expected load and its variability helps us provision appropriate capacity.

## 7. Independence and Conditional Probablity

Real-world phenomena rarely occur in isolation. Events influence each other in complex ways. Probability theory provides tools to reason about relationships between events, distinguishing between cases where events are unrelated (independent) and cases where knowledge of one event changes our beliefs about another (dependent).

Two events
`A` and 
`B` are **independent** if knowing whether one occurred provides no information about whether the other occurred. Mathematically, independence means that the probability of both events occurring equals the product of their individual probabilities:

$$P(A \cup B) = P(A) * P(B)$$


This reads as: *"the probability that both 
A and 
B occur equals the probability of 
A multiplied by the probability of 
B."*


Independence is a simplifying assumption that appears throughout statistical modeling. When we assume observations are independent, we can multiply their probabilities to compute joint likelihoods, enabling tractable inference in complex models. However, incorrectly assuming independence when dependence exists can lead to underestimating uncertainty and making overconfident predictions.


When events are dependent, we use **conditional probability** to express how the probability of one event changes given knowledge about another.The conditional probability of even $A$ given that event $B$ has occured is written $P(A|B)$ and defined as :

$$P(A|B) = P(A \cap B) / P(B) $$

This formula says that the conditional probability equals the probability of both events occurring divided by the probability of the conditioning event.
Rearranging this definition gives us the **multiplication rule**:

    $$P(A \cap B) = P(A|B) * P(B)$$

Conditional probability forms the basis for **Bayes' theorem**, one of the most important results in probability theory and the foundation of Bayesian statistical inference:
 
 $$  P(A|B)  = P(B|A) * P(A) / P(B)$$


Bayes' theorem allows us to "invert" conditional probabilities, computing 
$P(A∣B)$ from $P(B∣A)$. This inversion has many applications: in medical diagnosis, we might know the probability of a positive test given a disease, but we really want the probability of the disease given a positive test. In spam filtering, we observe words in an email and want to compute the probability the email is spam. In machine learning, we observe data and want to infer the probability of different model parameters.


**Example: Conditional Probability and Bayes' Theorem**

**Question**: If someone tests positive, what is the probability they have the disease?





> Notation: **¬ means 'not'**
> In probability and logic, 
`¬` (called "negation" or "not") denotes the complement of an event. So 
`¬D` means "not D" or "the event D does not occur." In our medical example, 
`¬D` represents a person who does not have the disease. You may also see this written as 
$D^{c}$ , $D{`}$ in different textbooks. They all mean the same thing: the complement of event D.


Interpretation: Even with a positive test result, the probability of having the disease is only about 8.8%! This counterintuitive result occurs because:

- The disease is rare (low prior probability)
- The false positive rate (10%) applied to the large healthy population generates many false positives
- These false positives "dilute" the true positives

This phenomenon is called the **base rate fallacy** and highlights why understanding conditional probability is essential for interpreting medical tests, machine learning classifiers, and any diagnostic system.



> Understanding **Sensitivity** and **Specificity**
> **Sensitivity** (also called the true positive rate) measures how well a test detects the condition when it is present. A sensitivity of 95% means that if 100 people with the disease are tested, 95 will correctly receive a positive result.
> **Specificity** (also called the true negative rate) measures how well a test correctly identifies people without the condition. A specificity of 90% means that if 100 healthy people are tested, 90 will correctly receive a negative result, but 10 will receive a false positive.

Despite the test appearing highly accurate (95% sensitivity, 90% specificity), the posterior probability of having the disease given a positive test is only about 8.8%. This counterintuitive result occurs because the disease is rare (1% prevalence). With 99 healthy people for every 1 sick person, even a 10% false positive rate generates many more false positives than true positives. This example illustrates why screening tests for rare conditions often require confirmation with follow-up testing.
